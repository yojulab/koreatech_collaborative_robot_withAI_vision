<h3><b>1. 라이브러리 가져오기 

```python
from pymycobot import MechArm
import time
```

<h3><b>2. MechArm 객체 생성하기 

```python
mc = MechArm("com9", 115200)
```

<h3><b>3. 움직임 관련 함수 생성하기 

dance 함수 생성하기

```python
def dance():
    init_pos = [0.17, -33.39, -12.56, -0.61, 43.94, 0.43]
    i = 0
    for i in range(3):
        mc.send_angles(init_pos,50)
        time.sleep(1)
        # print(mc.get_angles())
        mc.send_angle(4,40,50)
        time.sleep(0.1)
        mc.send_angle(6,40,50)
        time.sleep(1)
        mc.send_angles(init_pos,50)
        time.sleep(1)
        mc.send_angle(4,-40,50)
        time.sleep(0.1)
        mc.send_angle(6,-40,50)
        time.sleep(1)
        mc.send_angle(1,40,50)
        time.sleep(1)
        mc.send_angles([-90.7, 28.91, -163.3, 1.66, -47.81, 2.02],50)
        time.sleep(2)
        mc.send_angle(1,90,50)
        time.sleep(3)
        mc.send_angle(3,0,50)
        time.sleep(0.1)
        mc.send_angle(5,55,50)
        time.sleep(2)
        mc.send_angles([0.52, -10.37, -5.97, -1.14, 33.83, 1.75],50)
        time.sleep(1)
        mc.send_angle(2,-73.38,50)
        time.sleep(0.1)
        mc.send_angle(3,40,50)
        time.sleep(0.1)
        mc.send_angle(4,0,50)
        time.sleep(0.1)
        mc.send_angle(5,65,50)
        time.sleep(0.1)
    mc.send_angles(init_pos, 50)
```

thank_you 함수 생성하기 

```python
def thank_you():
    init_pos = [0, 0, 0, 0, 0, 0]
    up = [-4.92, -10.81, -33.39, -1.49, 36.03, -0.43]
    down = [-4.92, -37.35, 50.36, -2.37, 35.94, -0.43]

    i = 0
    mc.send_angles(init_pos, 50)
    time.sleep(1)
    for i in range(4):
        mc.send_angles(up, 40)
        time.sleep(2)
        mc.send_angles(down, 40)
        time.sleep(2)
        i += 1
    mc.send_angles(init_pos, 50)
```

<h3><b>4. 딕셔너리에 명령-함수 매핑하여 저장하기 

```python
commands = {
    "thank you" : thank_you,
    "dance" : dance,
    "0" : lambda : mc.send_angles([0, 0, 0, 0, 0, 0], 50),

}
```

<h3><b>5. 

```python
while True:
    order = input("움직일 모터 번호를 입력해주세요 (q를 누르면 종료) \n[ thank you, dance, 0, q ]: ")

    if order == "q":
        print("프로그램을 종료합니다.")
        break
    elif order in commands:
        commands[order]() # 입력된 명령에 해당하는 함수 실행 의미
        print("다시 원위치로 이동합니다.")
        time.sleep(2)
    else:
        print("잘못 입력하였습니다. 다시 입력해주세요.")
```
