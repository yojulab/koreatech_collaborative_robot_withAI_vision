<h1><b><font color=#6799FF>A. 특정 색상만 추출하기 </h1></b></font color>
<font size=2>파란색 범위를 설정하여 해당 값에 포함되는 객체만 출력하게끔 하고자 한다. 

<h5><b>1. 필요 라이브러리 가져오기 

```python
import cv2
import numpy as np
```

<h5><b>2. 색상 인식 관련 함수 생성하기 

```python
def color_detection(frame):
    # BGR을 HSV 형식으로 변환
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # 색상 범위 설정 (파랑색)
    lower_blue = np.array([110, 50, 50])
    upper_blue = np.array([130, 255, 255])

    # 이미지에서 파란색 객체를 찾기 위한 임계값
    mask = cv2.inRange(hsv_frame, lower_blue, upper_blue)

    # 원본 이미지에서 파란색 객체만 추출
    blue_objects = cv2.bitwise_and(frame, frame, mask=mask)

    return blue_objects
```

<h5><b>3. 카메라 활성화 시키기 

```python
# 웹캠 열기
cap = cv2.VideoCapture(0)
```

<h5><b>4. 실시간으로 특정 색상인식 확인하기 

```python
while True:
    # 프레임 읽기
    ret, frame = cap.read()
    if not ret:
        break

    # 색상 인식 수행
    result_frame = color_detection(frame)

    # 결과 보여주기
    cv2.imshow('Color Detection', result_frame)

    # 'q' 키를 누르면 종료
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 웹캠 해제 및 창 닫기
cap.release()
cv2.destroyAllWindows()
```

<h1><b><font color=#6799FF>B. 특정 색상 인식 후 윤곽선으로 표시하기 </h1></b></font color>
<font size=2>특정 색상 범위에 해당하는 객체를 찾아 윤곽선을 그리고 해당 색상의 이름을 표시하고자 한다. 

```python
import cv2
import numpy as np

def color_detection(frame, color_name, lower_color, upper_color):
    # BGR을 HSV 형식으로 변환
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # 색상 범위 설정
    lower_color = np.array(lower_color)
    upper_color = np.array(upper_color)

    # 이미지에서 해당 색상 객체를 찾기 위한 임계값
    mask = cv2.inRange(hsv_frame, lower_color, upper_color)

    # 원본 이미지에서 해당 색상 객체만 추출
    color_objects = cv2.bitwise_and(frame, frame, mask=mask)

    # 윤곽선 검출
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # 윤곽선 그리기 및 색상 이름 출력
    for contour in contours:
        # 윤곽선을 둘러싸는 최소한의 사각형 구하기
        x, y, w, h = cv2.boundingRect(contour)

        # 윤곽선 그리기
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # 색상 이름 출력
        cv2.putText(frame, color_name, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    return frame

# 색상 범위 설정 (파랑색)
lower_blue = [110, 50, 50]
upper_blue = [130, 255, 255]

# 카메라 열기
cap = cv2.VideoCapture(0)

while True:
    # 프레임 읽기
    ret, frame = cap.read()
    if not ret:
        break

    # 색상 인식 및 윤곽선 그리기
    frame_with_color = color_detection(frame, "Blue", lower_blue, upper_blue)

    # 결과 프레임 출력
    cv2.imshow('Color Detection', frame_with_color)

    # 종료 조건
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 카메라 해제 및 창 닫기
cap.release()
cv2.destroyAllWindows()
```

<h5><b>3. 색상 감지 및 윤곽선 검출 함수 생성하기</h5></b>
<font size=2>색상 감지 후, 그에 알맞게 로봇이 움직여야 하기 때문에 인식되는 부분이 너무 많으면 오류가 날 수 있다. 
<br>이를 방지하기 위해 출력될 수 있는 최소 바운딩 박스 크기를 지정한다. 

```python
# 색상 감지 및 윤곽선 검출 함수
def color_detection(frame, color_name, lower_color, upper_color, min_box_size):
    # BGR을 HSV 형식으로 변환
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # 색상 범위 설정
    lower_color = np.array(lower_color)
    upper_color = np.array(upper_color)

    # 이미지에서 해당 색상 객체를 찾기 위한 임계값
    mask = cv2.inRange(hsv_frame, lower_color, upper_color)

    # 원본 이미지에서 해당 색상 객체만 추출
    color_objects = cv2.bitwise_and(frame, frame, mask=mask)

    # 윤곽선 검출
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # 윤곽선 그리기 및 색상 이름 출력
    for contour in contours:
        # 윤곽선을 둘러싸는 최소한의 사각형 구하기
        x, y, w, h = cv2.boundingRect(contour)

        # 최소 바운딩 박스 크기 제어
        if w < min_box_size or h < min_box_size:
            continue  # 일정 크기 미만인 경우 윤곽선 무시

        # 윤곽선 그리기
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # 색상 이름 출력
        cv2.putText(frame, color_name, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

        # 로봇 팔 움직이기
        move_mechArm(color_name)  # 색상 이름으로 움직임 제어

    return frame
```

<h5><b>4. 색상 범위 설정 및 박스 최소 크기 설정하기</h5></b>

```python
# 색상 범위 설정 (파랑색)
lower_blue = [110, 50, 50]
upper_blue = [130, 255, 255]

# 색상 범위 설정 (빨간색)
lower_red = [0, 100, 100]
upper_red = [10, 255, 255]

# 색상 범위 설정 (초록색)
lower_green = [40, 100, 100]
upper_green = [80, 255, 255]

# 최소 바운딩 박스 크기 설정
min_box_size = 50  # 가로 또는 세로 길이가 50 픽셀 미만인 윤곽선은 무시
```
