```python
from pymycobot import MechArm
import time
```

```python
mc = MechArm("com9", 115200)
```

```python
mc.send_angles([0, 0, 0, 0, 0, 0], 50)
time.sleep(2)
mc.close()
```

```python
while True:
    motor_num = input("움직일 모터 번호를 입력해주세요 (q를 누르면 종료) : ")

    if motor_num == "q":
        print("프로그램을 종료합니다.")
        break

    mc = MechArm("com9", 115200)
    try:
        if motor_num in ["1", "2", "3", "4", "5", "6"]:
            mc.send_angle(int(motor_num), -40, 50)
            print("현재 J" + motor_num + "가 움직이고 있습니다.")
            time.sleep(2)
        elif motor_num == "0":
            mc.send_angles([0, 0, 0, 0, 0, 0], 50)
            print("다시 원위치로 이동합니다.")
            time.sleep(2)
        else:
            print("잘못 입력하였습니다. 다시 입력해주세요.")
    finally:
        mc.close()
```
