<h1><b> 데이터셋 분류하기

```python
import os
import shutil
import random
def split_dataset(dataset_dir, output_dir, train_ratio=0.8, val_ratio=0.1, test_ratio=0.1, seed=1337): 
    random.seed(seed)
    
    #이미지와 레이블 파일 디렉토리
    images_dir = os.path.join(dataset_dir, 'images')
    labels_dir = os.path.join(dataset_dir, 'labels')
    
    # 이미지 파일 목록 가져오기 (텍스트 파일은 이미지 파일과 동일한 이름으로 매칭)
    image_files = [f for f in os.listdir(images_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    # 파일무작위로 섞기
    random.shuffle(image_files)
    # 각 비율에 맞게 데이터셋 나누기
    total_files = len(image_files)
    train_split = int(total_files * train_ratio)
    val_split = int(total_files * (train_ratio + val_ratio))
    
    train_files = image_files[:train_split]
    val_files = image_files[train_split:val_split] 
    test_files = image_files[val_split:]
    
    # 출력 폴더 생성 (images labels 폴더 포함) 
    train_images_dir= os.path.join(output_dir, 'train', "images") 
    train_labels_dir = os.path.join(output_dir, 'train', 'labels') 
    val_images_dir = os.path.join(output_dir, 'val', 'images') 
    val_labels_dir = os.path.join(output_dir, 'val', 'labels') 
    test_images_dir = os.path.join(output_dir, 'test', "images") 
    test_labels_dir = os.path.join(output_dir, 'test', 'labels')
    
    os.makedirs(train_images_dir, exist_ok=True)
    os.makedirs(train_labels_dir, exist_ok=True)
    os.makedirs(val_images_dir, exist_ok=True)
    os.makedirs(val_labels_dir, exist_ok=True)
    os.makedirs(test_images_dir, exist_ok=True)
    os.makedirs(test_labels_dir, exist_ok=True)
    
    # # 이미지 파일과 레이블 파일을 각각의 폴더로 복사
    # for file in train_files:
    #     image_src = os.path.join(images_dir, file)
    #     label_src = os.path.join(labels_dir, file.rsplit('.', 1)[0] + '.txt') 
    #     shutil.copy(image_src, train_images_dir) 
    #     shutil.copy(label_src, train_labels_dir)
    # for file in val_files:
    #     image_src= os.path.join(images_dir, file)
    #     label_src = os.path.join(labels_dir, file.rsplit('.', 1)[0] + '.txt') 
    #     shutil.copy(image_src, val_images_dir) 
    #     shutil.copy(label_src, val_labels_dir)
    # for file in test_files:
    #     image_src = os.path.join(images_dir, file)
    #     label_src = os.path.join(labels_dir, file.rsplit('.', 1)[0] + '.txt') 
    #     shutil.copy(image_src, test_images_dir) 
    #     shutil.copy(label_src, test_labels_dir)
    
    print(f'Total files: {total_files}')
    print(f'Train files: {len(train_files)}') 
    print(f'Validation files: {len(val_files)}') 
    print(f'Test files: {len(test_files)}')
    
    #예시 사용법
split_dataset('dataset', 'output', train_ratio=0.8, val_ratio=0.1, test_ratio=0.1)
```
